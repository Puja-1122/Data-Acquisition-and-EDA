```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("cleaned_titanic.csv")

df.head()
```

```python
print("Rows and Columns:", df.shape)

df.info()

df.describe()
```

## **Understanding Passenger Survival Patterns on the Titanic Through Data Visualization**

# Objective

The objective of this project is to use advanced data visualization techniques to analyze passenger characteristics and identify the factors influencing survival on the Titanic. Through a series of visualizations, this report presents a clear data story that can be easily understood by both technical and non-technical audiences.

### Visualization 1-Survival Distribution

```python
plt.figure(figsize=(6,5))

sns.countplot(x='Survived', data=df, palette='Set2')

plt.title("Passenger Survival Distribution", fontsize=16)

plt.xlabel("Survival")

plt.ylabel("Number of Passengers")

plt.show()
```

### Story

This visualization shows the overall number of passengers who survived and those who did not. It provides a starting point for understanding the survival patterns before exploring other influencing factors.

**Visualization 2-Survival by Gender**

```python
plt.figure(figsize=(7,5))

sns.countplot(x='Sex',
              hue='Survived',
              data=df,
              palette='viridis')

plt.title("Survival by Gender")

plt.show()
```

Story

This chart demonstrates that female passengers had a much higher survival rate than male passengers, reflecting the historical evacuation policy of prioritizing women and children.

**Visualization 3-Age Distribution**

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"],
             bins=30,
             kde=True,
             color="steelblue")

plt.title("Age Distribution of Passengers")

plt.xlabel("Age")

plt.show()
```

Story

The histogram reveals that most passengers were adults between 20 and 40 years old, while fewer passengers belonged to younger and older age groups.

**Visualization 4-Passenger Class vs Survival**

```python
plt.figure(figsize=(8,5))

sns.countplot(x='Pclass',
              hue='Survived',
              data=df,
              palette='Set1')

plt.title("Passenger Class and Survival")

plt.show()
```

Story

Passengers traveling in first class generally had a higher survival rate than those in lower classes, indicating that socioeconomic status influenced survival outcomes.

**Visualization 5-Fare Distribution**

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=df["Fare"],
            color="orange")

plt.title("Distribution of Ticket Fare")

plt.show()
```

Story

The box plot highlights several high-fare outliers, suggesting that some passengers paid significantly more than the majority, likely for premium accommodations.

**Data Storytelling**

The visualizations above can help to understand the factors that affected people’s survival. First of all, the charts demonstrate that gender significantly contributed to survival rates as female passengers had a higher chance of surviving than male ones. Second, the survival probability depended on the passengers’ class and the fare they paid. It implies that the socioeconomic status of an individual influenced their likelihood of being safe during the ship crash. The age distribution indicates that most of the passengers were adults, while the heatmap illustrates the relationship between different numeric variables. Altogether, the graphs above demonstrate the connection between various factors and passengers’ survival and help visualize the entire dataset.
